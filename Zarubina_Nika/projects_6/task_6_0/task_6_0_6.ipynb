```python
import pandas as pd
df = pd.read_csv('wild_boars.csv')
males = df[df['gender'] == 'Male']      
females = df[df['gender'] == 'Female'] 
male_length = males['length_cm']

male_q1 = male_length.quantile(0.25)
male_q3 = male_length.quantile(0.75)
male_iqr = male_q3 - male_q1

female_length = females['length_cm']
female_q1 = female_length.quantile(0.25)
female_q3 = female_length.quantile(0.75)
female_iqr = female_q3 - female_q1


result = pd.DataFrame({'Пол': ['Самец', 'Самка'], 'Q1 (25-й перцентиль)': [male_q1, female_q1], 'Q3 (75-й перцентиль)': [male_q3, female_q3], 'Межквартильный размах (IQR)': [male_iqr, female_iqr]})

# Сохраняем результат в CSV файл
result.to_csv('iqr_by_gender.csv', index=False)

# Выводим на экран для проверки
print("Результаты расчёта межквартильного размаха длины тела:")
print(result)
print("\nФайл 'iqr_by_gender.csv' успешно сохранён!")
```
