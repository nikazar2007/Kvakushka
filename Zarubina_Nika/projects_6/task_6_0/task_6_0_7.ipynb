```python
import pandas as pd
df = pd.read_csv('wild_boars.csv')
numbers = df.select_dtypes(include=['number'])
variance = numbers.var()
std_dev = numbers.std()
cv = (std_dev / numbers.mean()) * 100

# Скл
result = pd.DataFrame({
    'Дисперсия': variance,
    'Стандартное_отклонение': std_dev,
    'Коэффициент_вариации_%': cv
})

# Сохраняем в файл
result.to_csv('variability_stats.csv')

# Выводим на экран
print(result)
```
