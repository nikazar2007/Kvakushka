```python
import pandas as pd
df = pd.read_csv('wild_boars.csv')
males = df[df['gender'] == 'Male']
females = df[df['gender'] == 'Female']

male_tusk = males['tusk_length_cm'].dropna()
female_tusk = females['tusk_length_cm'].dropna()

male_cv = (male_tusk.std() / male_tusk.mean()) * 100
female_cv = (female_tusk.std() / female_tusk.mean()) * 100
result = pd.DataFrame({'Пол': ['Самец', 'Самка'], 'Коэффициент вариации длины клыков (%)': [male_cv, female_cv]})
result.to_csv('tusk_cv_by_gender.csv', index=False)


print(result)
print("\nФайл 'tusk_cv_by_gender.csv' сохранён!")
```
